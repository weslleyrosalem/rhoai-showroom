```python
from pathlib import Path
import sys
repo = Path.cwd()
if repo.name == "notebooks": repo = repo.parent
sys.path.insert(0, str(repo / "apps" / "aurora-rag"))
from rag import Retriever
retriever = Retriever(repo / "data" / "documents")
```

```python
question = "Who must approve a replenishment proposal totaling 6000?"
[(d["document_id"], d["score"]) for d in retriever.retrieve(question)]
```

```python
import json, urllib.request
payload=json.dumps({"question":"Should I replenish AS-001? Explain the policy and required approval."}).encode()
request=urllib.request.Request("http://aurora-rag:8080/ask",data=payload,headers={"Content-Type":"application/json"})
with urllib.request.urlopen(request, timeout=150) as response: result=json.load(response)
print(result["answer"])
print(result["sources"], result.get("trace_id"))
```
